```python
df['Admission_Type'].value_counts().plot(kind='bar')
```

```python
Billing_Summary=(df.groupby(['Medical_Condition','Admission_Type'])['Billing_Amount'].sum().unstack())
```

```python
Billing_Summary.plot(kind='bar',stacked=True,figsize=(12,6),edgecolor='black')
plt.title('Total Billing Amount by Medical Condition and Admission Type')
plt.xlabel('Medical Condition',fontsize=12)
plt.ylabel('Total Billing Amount',fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Admission Type')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(data=df,x='Medical_Condition',y='Billing_Amount')
plt.title('Billing Amount Distribution by Medical Condition',fontsize=16)
plt.xlabel('Medical Condition')
plt.ylabel('Billing Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Date_of_Admission']=pd.to_datetime(
    df['Date_of_Admission']
    )
```

```python
monthly_admission=(
    df.set_index('Date_of_Admission')

    .resample('ME')
    .size()
)
print(monthly_admission)
```

```python
plt.figure(figsize=(14,6))
plt.plot(
    monthly_admission.index,
    monthly_admission.values,
    marker='o',
)
plt.title('Monthly Patient Admission',fontsize=16)
plt.xlabel('Month',fontsize=12)
plt.ylabel('Number of Admission,fontsize=12')
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.show()
```
